```python
regions = input("Enter regions: ").split()
colors = input("Enter colors: ").split()

graph = {}

for r in regions:
    graph[r] = input("Neighbours of "+r+": ").split()

solution = {}

def solve(i):
    if i == len(regions):
        return True

    r = regions[i]

    for c in colors:
        if all(solution.get(x) != c for x in graph[r]):
            solution[r] = c

            if solve(i+1):
                return True

            del solution[r]

    return False

solve(0)
print("Solution:",solution)
```
