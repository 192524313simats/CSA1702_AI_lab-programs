```python
teachers = {
    "Ravi": "Computer",
    "Anu": "Maths"
}

students = {
    "Kumar": "Computer",
    "Raj": "Maths"
}

teacher = input("Enter teacher name: ")

if teacher in teachers:
    subject = teachers[teacher]

    for i in students:
        if students[i] == subject:
            print("Student:", i)
else:
    print("Teacher not found")
```
