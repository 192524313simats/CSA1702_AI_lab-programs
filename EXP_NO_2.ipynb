```python
r = int(input("Enter rows: "))
c = int(input("Enter columns: "))

A = []
B = []

print("Enter Matrix A:")
for i in range(r):
    A.append(list(map(int, input().split())))

print("Enter Matrix B:")
for i in range(r):
    B.append(list(map(int, input().split())))

C = []

for i in range(r):
    row = []
    for j in range(c):
        row.append(A[i][j] + B[i][j])
    C.append(row)

print("Result:")
for i in range(r):
    print(C[i])
```
