```python
N = 8

def solve(row,pos):
    if row == N:
        print(pos)
        return True

    for col in range(N):
        if all(col != c and abs(col-c) != row-r
               for r,c in enumerate(pos)):
            if solve(row+1,pos+[col]):
                return True

    return False

solve(0,[])
```
