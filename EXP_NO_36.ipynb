```python
graph = {
    "A": [("B", 4), ("C", 2)],
    "B": [("D", 5)],
    "C": [("D", 1)],
    "D": []
}

start = input("Enter start node: ")
goal = input("Enter goal node: ")

open_list = [(0, start)]
visited = set()

while open_list:
    open_list.sort()
    cost, node = open_list.pop(0)

    print(node, end=" ")

    if node == goal:
        print("\nGoal reached")
        break

    if node in visited:
        continue

    visited.add(node)

    for next_node, value in graph[node]:
        if next_node not in visited:
            open_list.append((value, next_node))
```
