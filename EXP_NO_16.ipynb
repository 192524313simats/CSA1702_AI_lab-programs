```python
from collections import deque

g = {}

n = int(input("Number of nodes: "))

for i in range(n):
    x = input("Node: ")
    g[x] = input("Neighbours: ").split()

start = input("Start node: ")

q = deque([start])
seen = {start}

while q:
    x = q.popleft()
    print(x,end=" ")

    for y in g[x]:
        if y not in seen:
            seen.add(y)
            q.append(y)
```
