```python
r = int(input("Enter rows: "))
c = int(input("Enter columns: "))

A = []

for i in range(r):
    A.append(list(map(int, input().split())))

T = []

for i in range(c):
    row = []
    for j in range(r):
        row.append(A[j][i])
    T.append(row)

print("Transpose:")
for i in range(c):
    print(T[i])
```
