```python
facts = ["rain"]

rules = {
    "road_wet": "rain",
    "drive_slow": "road_wet"
}

goal = input("Enter goal: ")

def backward(goal):
    if goal in facts:
        return True

    if goal in rules:
        return backward(rules[goal])

    return False

if backward(goal):
    print("Goal can be proved")
else:
    print("Goal cannot be proved")
```
