```python
def ab(x,d,alpha,beta,maxi):

    if d==0 or not isinstance(x,list):
        return x

    if maxi:
        v=-999

        for i in x:
            v=max(v,ab(i,d-1,alpha,beta,False))
            alpha=max(alpha,v)

            if alpha>=beta:
                break

        return v

    else:
        v=999

        for i in x:
            v=min(v,ab(i,d-1,alpha,beta,True))
            beta=min(beta,v)

            if alpha>=beta:
                break

        return v

tree = eval(input("Enter game tree: "))
d = int(input("Enter depth: "))

print("Best value:",ab(tree,d,-999,999,True))
```
