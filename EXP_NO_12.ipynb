```python
from collections import deque

a = int(input("Jug 1 capacity: "))
b = int(input("Jug 2 capacity: "))
target = int(input("Target: "))

q = deque([((0,0),[])])
seen = {(0,0)}

while q:
    (x,y),path = q.popleft()

    if x == target or y == target:
        print("Solution:",path+[(x,y)])
        break

    states = [
        (a,y),(x,b),(0,y),(x,0),
        (x-min(x,b-y),y+min(x,b-y)),
        (x+min(y,a-x),y-min(y,a-x))
    ]

    for s in states:
        if s not in seen:
            seen.add(s)
            q.append((s,path+[(x,y)]))
```
