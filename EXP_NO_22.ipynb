```python
def minimax(x,maxi):
    if not isinstance(x,list):
        return x

    v = []

    for i in x:
        v.append(minimax(i,not maxi))

    if maxi:
        return max(v)
    else:
        return min(v)

tree = eval(input("Enter game tree: "))

print("Best value:",minimax(tree,True))
```
