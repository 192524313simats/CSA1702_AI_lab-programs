```python
male = ["Tom", "Jim", "Pam"]
female = ["Liz", "Ann", "Pat"]

parent = {
    "Tom": ["Liz", "Pam"],
    "Liz": ["Jim"],
    "Ann": ["Pam"],
    "Pam": ["Pat"]
}

name = input("Enter name: ")

print("Children:", parent.get(name, []))

for p in parent:
    if name in parent[p]:
        if p in male:
            print("Father:", p)
        else:
            print("Mother:", p)
```
