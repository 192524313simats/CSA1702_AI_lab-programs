```python
fever = input("Fever? yes/no: ")
cough = input("Cough? yes/no: ")
cold = input("Cold? yes/no: ")

if fever == "yes" and cough == "yes":
    print("Diagnosis: Flu")
elif fever == "yes" and cold == "yes":
    print("Diagnosis: Viral Fever")
elif cough == "yes":
    print("Diagnosis: Cough")
else:
    print("Diagnosis not found")
```
