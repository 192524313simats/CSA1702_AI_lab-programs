```python
monkey = input("Is monkey near box? yes/no: ")
box = input("Is box near banana? yes/no: ")
reach = input("Can monkey reach banana? yes/no: ")

if monkey == "yes" and box == "yes" and reach == "yes":
    print("Monkey gets banana")
else:
    print("Monkey cannot get banana")
```
