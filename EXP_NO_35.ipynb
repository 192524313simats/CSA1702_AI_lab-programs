```python
fruit = {
    "apple": "red",
    "banana": "yellow",
    "orange": "orange",
    "grapes": "green"
}

name = input("Enter fruit: ").lower()

if name in fruit:
    print("Color:", fruit[name])
else:
    print("Fruit not found")
```
