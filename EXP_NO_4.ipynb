```python
s = input("Enter sentence: ")
words = s.split()

for i in range(len(words)):
    for j in range(i + 1, len(words)):
        if words[i].lower() > words[j].lower():
            words[i], words[j] = words[j], words[i]

print("Sorted sentence:")

for i in range(len(words)):
    print(words[i], end=" ")
```
