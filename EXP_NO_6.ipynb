```python
L = list(map(int, input("Enter list: ").split()))

L.append(10)
print("After append:", L)

L.extend([20, 30])
print("After extend:", L)

L.insert(1, 50)
print("After insert:", L)

L.remove(50)
print("After remove:", L)

L.sort()
print("After sort:", L)

L.reverse()
print("After reverse:", L)

L.pop()
print("After pop:", L)
```
