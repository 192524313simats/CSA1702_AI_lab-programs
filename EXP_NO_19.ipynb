```python
import heapq

graph = {
'A':[('B',1),('C',3)],
'B':[('D',3)],
'C':[('D',1)],
'D':[]
}

h = {'A':3,'B':2,'C':1,'D':0}

start = input("Start: ")
goal = input("Goal: ")

q = [(h[start],0,start)]
seen = set()

while q:
    f,c,x = heapq.heappop(q)

    if x == goal:
        print("Cost:",c)
        break

    if x in seen:
        continue

    seen.add(x)

    for y,w in graph[x]:
        heapq.heappush(q,(c+w+h[y],c+w,y))
```
