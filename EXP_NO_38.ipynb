```python
facts = []

n = int(input("Enter number of facts: "))

for i in range(n):
    facts.append(input("Enter fact: "))

if "rain" in facts:
    facts.append("road_wet")

if "road_wet" in facts:
    facts.append("drive_slow")

print("Facts after forward chaining:")

for i in facts:
    print(i)
```
