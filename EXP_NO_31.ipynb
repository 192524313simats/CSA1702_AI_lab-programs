```python
birds = ["sparrow", "eagle", "penguin", "ostrich"]
flying = ["sparrow", "eagle"]

bird = input("Enter bird: ")

if bird in flying:
    print(bird, "can fly")
elif bird in birds:
    print(bird, "cannot fly")
else:
    print("Bird not found")
```
