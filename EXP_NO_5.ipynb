```python
L = list(map(int, input("Enter list elements: ").split()))

print("List =", L)
print("Length =", len(L))

print("Iteration:")
for i in range(len(L)):
    print(L[i])

x = int(input("Enter element to search: "))

if x in L:
    print("Element found")
else:
    print("Element not found")

print("First element =", L[0])
print("Slice =", L[1:])
```
