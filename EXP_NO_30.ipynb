```python
def hanoi(n, source, target, temp):
    if n == 1:
        print("Move disk from", source, "to", target)
        return

    hanoi(n-1, source, temp, target)
    print("Move disk from", source, "to", target)
    hanoi(n-1, temp, target, source)

n = int(input("Enter number of disks: "))

hanoi(n, "A", "C", "B")
```
