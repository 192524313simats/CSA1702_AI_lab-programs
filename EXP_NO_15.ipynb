```python
A = input("Room A: ")
B = input("Room B: ")
pos = input("Starting position: ")

room = {'A':A,'B':B}

while "Dirty" in room.values():
    if room[pos] == "Dirty":
        print("Cleaning",pos)
        room[pos] = "Clean"
    else:
        pos = 'B' if pos=='A' else 'A'

print("Both rooms are clean")
```
