```python
from collections import deque

q = deque([((3,3,0),[])])
seen = {(3,3,0)}

moves = [(1,0),(2,0),(0,1),(0,2),(1,1)]

def safe(m,c):
    return 0<=m<=3 and 0<=c<=3 and \
           (m==0 or m>=c) and \
           (3-m==0 or 3-m>=3-c)

while q:
    (m,c,b),path = q.popleft()

    if m==0 and c==0:
        print(path+[(m,c,b)])
        break

    for x,y in moves:
        if b==0:
            s=(m-x,c-y,1)
        else:
            s=(m+x,c+y,0)

        if safe(s[0],s[1]) and s not in seen:
            seen.add(s)
            q.append((s,path+[(m,c,b)]))
```
