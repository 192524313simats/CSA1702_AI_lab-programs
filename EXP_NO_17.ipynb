```python
g = {}

n = int(input("Number of nodes: "))

for i in range(n):
    x = input("Node: ")
    g[x] = input("Neighbours: ").split()

start = input("Start node: ")
seen = set()

def dfs(x):
    print(x,end=" ")
    seen.add(x)

    for y in g[x]:
        if y not in seen:
            dfs(y)

dfs(start)
```
